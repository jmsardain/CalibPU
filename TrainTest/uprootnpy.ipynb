```python
import uproot as up
import numpy as np
import pandas as pd
```

```python
tree_dict = {}
paths = ["/data/dsmith/data_for_test/user.cdelitzs.45207441._000199.mltree_cluster_calo.root"]
# paths = ["/data/dsmith/mc_for_test/MC20d_JZ0_putagging_snapshot.root", "/data/dsmith/mc_for_test/MC20d_JZ1_putagging_snapshot.root", 
#          "/data/dsmith/mc_for_test/MC20d_JZ2_putagging_snapshot.root", "/data/dsmith/mc_for_test/MC20d_JZ3_putagging_snapshot.root",
#          "/data/dsmith/mc_for_test/MC20d_JZ4_putagging_snapshot.root", "/data/dsmith/mc_for_test/MC20d_JZ5_putagging_snapshot.root",
#          "/data/dsmith/mc_for_test/MC20d_JZ6_putagging_snapshot.root"]
```

```python
for path in paths:
    with up.open(path) as file:
        df = pd.DataFrame()
        tree = file["ClusterTree"]
        # print(tree.keys())
        print(len(tree.keys()))
        for key in tree.keys():
            tree_dict[key] = tree[key].array()
        df = pd.DataFrame.from_dict(tree_dict)
        df["diffEta"] = df["clusterEta"]-df["jetRawEta"]
        df["zT"] = df["clusterE"]/df["jetRawE"]
        df["zL"] = df["clusterE"]/df["jetRawE"]
        df["zRel"] = df["clusterE"]/df["jetRawE"]

columns = ['clusterE',   ## 0
        'clusterEta', ## 1
        'cluster_time', ## 2
        'cluster_CENTER_LAMBDA', ## 3
        'cluster_CENTER_MAG', ## 4
        'cluster_ENG_FRAC_EM', ## 5  
        'cluster_FIRST_ENG_DENS', ## 6
        'cluster_LATERAL', ## 7
        'cluster_LONGITUDINAL', ## 8
        'cluster_PTD', ## 9
        'cluster_ISOLATION', ## 10
        'cluster_SECOND_TIME', ## 11
        'cluster_SIGNIFICANCE', ## 12
        'nPrimVtx', ## 13
        'avgMu', ## 14
        'clusterPhi', ## 15
        'diffEta', ## 16
        'zT', 'zL', 'zRel', ## 17,18,19
        'cluster_nCells' ## 20
          ]
        
df2 = pd.DataFrame()
pd.concat([df2, df])
for column in columns:
    df2[column] = df[column]
    print(column+':', len(df2[column]))
    print("mean:", np.mean(df2[column]), "\nstd:", np.std(df2[column]))
df2.to_numpy()
np.save("/data/dsmith/mc_for_test/testfile.npy", df2)
```
