```python
import numpy as np
import pandas as pd
import uproot
```

```python
def transform(variable):
    if variable=="clusterE": mean, std =  0.769285 , 1.66154
    if variable=="cluster_CENTER_LAMBDA": mean, std =  5.48896 , 1.11419
    if variable=="cluster_FIRST_ENG_DENS": mean, std =  -14.1986 , 1.68766
    if variable=="cluster_SECOND_TIME": mean, std =  -1.27973 , 6.68972
    if variable=="cluster_SIGNIFICANCE": mean, std =  1.39433 , 1.1059
    if variable=="cluster_CENTER_MAG": mean, std =  7.98977 , 0.372179
    return [mean, std]
```

```python
def apply_save_log(x):

    #########
    epsilon = 1e-10
    #########

    minimum = x.min()
    if x.min() <= 0:
        x = x - x.min() + epsilon
    else:
        minimum = 0
        epsilon = 0

    return np.log(x)
```

```python
feature_branches = [
        'clusterE',   ## 0
        'clusterEta', ## 1
        'cluster_time', ## 2
        'cluster_CENTER_LAMBDA', ## 3
        'cluster_CENTER_MAG', ## 4
        'cluster_ENG_FRAC_EM', ## 5  
        'cluster_FIRST_ENG_DENS', ## 6
        'cluster_LATERAL', ## 7
        'cluster_LONGITUDINAL', ## 8
        'cluster_PTD', ## 9
        'cluster_ISOLATION', ## 10
        'cluster_SECOND_TIME', ## 11
        'cluster_SIGNIFICANCE', ## 12
        'nPrimVtx', ## 13
        'avgMu', ## 14
        'clusterPhi', ## 15
        # 'diffEta', ## 16
        # 'zT', 'zL', 'zRel', ## 17,18,19
        'cluster_nCells' ## 20
    ]
```

```python
tree_dict = {}
with uproot.open('/data/dsmith/data_for_test/user.cdelitzs.45207441._000199.mltree_cluster_calo.root') as file:
    df = pd.DataFrame()
    tree = file["ClusterTree"]
    for key in tree.keys():
        tree_dict[key] = tree[key].array()
    df = pd.DataFrame.from_dict(tree_dict)
    df2 = df[feature_branches]

    df2["diffEta"] = df["clusterEta"] = df["jetRawEta"]
    df2["zT"] = df["clusterE"]/df["jetRawE"]
    df2["zL"] = df["clusterE"]/df["jetRawE"]
    df2["zRel"] = df["clusterE"]/df["jetRawE"]
    
```

```python
variable_list = ["clusterE", "cluster_CENTER_LAMBDA", "cluster_FIRST_ENG_DENS", "cluster_SECOND_TIME", "cluster_SIGNIFICANCE", "cluster_CENTER_MAG"]
for variable in variable_list:
    df2[variable] = (apply_save_log(df[variable]) - transform(variable)[0])/transform(variable)[1]
    
```

```python
array = df2.to_numpy()
np.save("/data/dsmith/data_for_test/test_data.npy", array)
```
